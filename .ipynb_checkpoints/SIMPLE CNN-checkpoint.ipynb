```python
from torchvision.datasets import ImageFolder
```

```python
transform = transforms.Compose([
    transforms.Grayscale(num_output_channels=1),
    transforms.Resize((128, 128)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.5], std=[0.5])
])
```

```python
train_dataset=ImageFolder(root='D:\\dataset\\train', transform=transform)
test_dataset=ImageFolder(root='D:\\dataset\\test', transform=transform)
```

```python
train_loader=DataLoader(train_dataset,batch_size=32,shuffle=True)
```

```python
test_loader=DataLoader(test_dataset,batch_size=32,shuffle=False)
```

```python
print(train_dataset.classes)
print(f"Train size: {len(train_dataset)}, Test size: {len(test_dataset)}")
```

```python
class SimpleCNN(nn.Module):

    def __init__(self):
        super(SimpleCNN, self).__init__()

        # First Convolution Block
        self.conv1 = nn.Conv2d(
            in_channels=1,
            out_channels=32,
            kernel_size=3,
            padding=1
        )

        # Pooling Layer
        self.pool = nn.MaxPool2d(2, 2)

        # Second Convolution Block
        self.conv2 = nn.Conv2d(
            32,
            64,
            kernel_size=3,
            padding=1
        )

        # Third Convolution Block
        self.conv3 = nn.Conv2d(
            64,
            128,
            kernel_size=3,
            padding=1
        )

        # Fully Connected Layers
        self.fc1 = nn.Linear(128 * 16 * 16, 128)

        self.fc2 = nn.Linear(128, 2)

    def forward(self, x):

        x = self.pool(torch.relu(self.conv1(x)))

        x = self.pool(torch.relu(self.conv2(x)))

        x = self.pool(torch.relu(self.conv3(x)))

        # Flatten
        x = x.view(x.size(0), -1)

        x = torch.relu(self.fc1(x))

        x = self.fc2(x)

        return x
```

```python
model = SimpleCNN()

print(model)
```

```python
criterion = nn.CrossEntropyLoss()
```

```python
optimizer = optim.Adam(
    model.parameters(),
    lr=0.001
)
```

```python
num_epochs = 10

for epoch in range(num_epochs):

    running_loss = 0.0

    for images, labels in train_loader:

        # Forward pass
        outputs = model(images)

        loss = criterion(outputs, labels)

        # Clear old gradients
        optimizer.zero_grad()

        # Backpropagation
        loss.backward()

        # Update weights
        optimizer.step()

        running_loss += loss.item()
        
    #epoch_loss = running_loss / len(train_loader)

    print(f"Epoch [{epoch+1}/{num_epochs}], Loss: {running_loss:.4f}")
```

```python
correct = 0
total = 0

model.eval()

with torch.no_grad():

    for images, labels in test_loader:

        outputs = model(images)

        _, predicted = torch.max(outputs, 1)

        total += labels.size(0)

        correct += (predicted == labels).sum().item()

accuracy = 100 * correct / total

print(f"Test Accuracy: {accuracy:.2f}%")
```

```python

image_path = "C:\\Users\\senku\\OneDrive\\Desktop\\normal chest_non_dataset.jpeg"

image_name = os.path.basename(image_path)

# Load original image
original_image = Image.open(image_path).convert("L")

# Transform for model
input_tensor = transform(original_image).unsqueeze(0)

# Prediction
model.eval()

with torch.no_grad():

    output = model(input_tensor)

    _, predicted = torch.max(output, 1)

predicted_class = classes[predicted.item()]

print("Prediction:", predicted_class)

# ------------------------
# Grad-CAM
# ------------------------

target_layer = model.conv3

cam = GradCAM(
    model=model,
    target_layers=[target_layer]
)

# Generate CAM
grayscale_cam = cam(input_tensor=input_tensor)

grayscale_cam = grayscale_cam[0]

# Prepare image for overlay
rgb_image = np.array(
    original_image.resize((128,128))
) / 255.0

# Convert grayscale to RGB
rgb_image = np.stack([rgb_image]*3, axis=-1)

# Overlay heatmap
visualization = show_cam_on_image(
    rgb_image,
    grayscale_cam,
    use_rgb=True
)

# Display
plt.figure(figsize=(6,6))

plt.imshow(visualization)

plt.title(f"Prediction: {predicted_class}")


plt.axis("off")

save_name = f"sample_output_{image_name}"
plt.savefig(
    save_name,
    bbox_inches='tight'
)

plt.show()
```
